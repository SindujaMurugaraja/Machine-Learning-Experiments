```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, StandardScaler
```

```python
df = pd.read_csv("/content/student_performance_dirty.csv")
df.head()
```

```python
print("Shape of dataset:", df.shape)
print("\nColumn names:")
print(df.columns)
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df["Hours Studied"] = df["Hours Studied"].fillna(df["Hours Studied"].mean())
df["Sleep Hours"] = df["Sleep Hours"].fillna(df["Sleep Hours"].mean())
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
df.duplicated().sum()
```

```python
print(df.dtypes)
```

```python
le = LabelEncoder()

df["Extracurricular Activities"] = le.fit_transform(
    df["Extracurricular Activities"]
)

df["Notes"] = le.fit_transform(
    df["Notes"]
)
```

```python
df.head()
```

```python
print(df.columns.tolist())
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

numerical_columns = [
    "Hours Studied",
    "Previous Scores",
    "Sleep Hours",
    "Sample Question Papers Practiced"
]

df[numerical_columns] = scaler.fit_transform(
    df[numerical_columns]
)

df.head()
```

```python
print("Final Shape:", df.shape)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nData Types:")
print(df.dtypes)
```

```python
df.to_csv("student_preprocessed.csv", index=False)
```
